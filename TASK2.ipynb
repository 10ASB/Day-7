# Task 2 — Decision Tree Visualization and Overfitting

Train and visualize a decision tree using `plot_tree`. Compare a maximum depth of 20 with a maximum depth of 4.

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.metrics import accuracy_score
import matplotlib.pyplot as plt
```

## Load and split the dataset

```python
data = load_iris()
X = data.data
y = data.target
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)
```

## Train trees with depth 20 and depth 4

```python
tree_20 = DecisionTreeClassifier(max_depth=20, random_state=42)
tree_20.fit(X_train, y_train)
tree_4 = DecisionTreeClassifier(max_depth=4, random_state=42)
tree_4.fit(X_train, y_train)
```

## Predictions

```python
pred_20 = tree_20.predict(X_test)
pred_4 = tree_4.predict(X_test)
```

## Accuracy

```python
print("Depth 20")
print("Training Accuracy:", round(tree_20.score(X_train, y_train), 4))
print("Testing Accuracy :", round(accuracy_score(y_test, pred_20), 4))
print("\nDepth 4")
print("Training Accuracy:", round(tree_4.score(X_train, y_train), 4))
print("Testing Accuracy :", round(accuracy_score(y_test, pred_4), 4))
```

## Visualize trees

```python
plt.figure(figsize=(14, 10))
plot_tree(tree_20, feature_names=data.feature_names,
          class_names=data.target_names, filled=True)
plt.title("Decision Tree - Depth 20")
plt.show()

plt.figure(figsize=(18, 10))
plot_tree(tree_4, feature_names=data.feature_names,
          class_names=data.target_names, filled=True)
plt.title("Decision Tree - Depth 4")
plt.show()
```
