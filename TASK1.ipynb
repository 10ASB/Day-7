# Task 1 — Entropy and Information Gain

Compute one entropy/information-gain split by hand on paper, then verify it in code.

## Sample data

```python
import math

a = ["Yes", "Yes", "Yes", "YES", "No", "No", "No", "No"]

# Function to calculate entropy
def entropy(labels):
    total = len(labels)
    counts = {}
    for label in labels:
        counts[label] = counts.get(label, 0) + 1
    ent = 0
    for count in counts.values():
        p = count / total
        ent -= p * math.log2(p)
    return ent
```

## Split based on the feature

```python
# Parent entropy
parent_entropy = entropy(a)

# Split based on feature
left = ["Yes", "Yes", "Yes", "YES"]
right = ["No", "No", "No", "No"]
```

## Weighted entropy after split

```python
total = len(left) + len(right)
weighted_entropy = (
    (len(left) / total) * entropy(left)
    + (len(right) / total) * entropy(right)
)
```

## Information gain

```python
information_gain = parent_entropy - weighted_entropy
print("Parent Entropy:", round(parent_entropy, 4))
print("Left Entropy:", round(entropy(left), 4))
print("Right Entropy:", round(entropy(right), 4))
print("Weighted Entropy:", round(weighted_entropy, 4))
print("Information Gain:", round(information_gain, 4))
```
