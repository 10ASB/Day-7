# Task 3 — Random Forest, Feature Importance and Tree Comparison

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.inspection import permutation_importance
from sklearn.metrics import accuracy_score
import pandas as pd
```

## Load dataset and split data

```python
data = load_iris()
X = data.data
y = data.target
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=100
)
```

## Train Random Forest and check accuracy

```python
rf = RandomForestClassifier(n_estimators=100, random_state=78, oob_score=True)
rf.fit(X_train, y_train)
train_accuracy = rf.score(X_train, y_train)
test_accuracy = rf.score(X_test, y_test)
print("\nRandom Forest")
print("\nTraining Accuracy:", round(train_accuracy, 4))
print("Testing Accuracy:", round(test_accuracy, 4))
print("OOB Score:", round(rf.oob_score_, 4))
```

## Gini feature importance

```python
gini_importance = pd.DataFrame({
    "Feature": data.feature_names,
    "Gini Importance": rf.feature_importances_
})
print("\nGini Feature Importance:")
print(gini_importance.sort_values("Gini Importance", ascending=False))
```

## Permutation importance

```python
perm = permutation_importance(rf, X_test, y_test, n_repeats=10, random_state=50)
perm_importance = pd.DataFrame({
    "Feature": data.feature_names,
    "Permutation Importance": perm.importances_mean
})
print("\nPermutation Importance:")
print(perm_importance.sort_values("Permutation Importance", ascending=False))
```

## Compare with a Decision Tree

```python
data = load_iris()
X = data.data
y = data.target
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)
```

## Train Decision Tree and Random Forest, then predict

```python
tree = DecisionTreeClassifier(max_depth=4, random_state=52)
tree.fit(X_train, y_train)
forest = RandomForestClassifier(n_estimators=2000, random_state=69)
forest.fit(X_train, y_train)
tree_pred = tree.predict(X_test)
forest_pred = forest.predict(X_test)
```

## Accuracy comparison

```python
tree_accuracy = accuracy_score(y_test, tree_pred)
forest_accuracy = accuracy_score(y_test, forest_pred)
print("\nMODEL COMPARISON")
print("\nDecision Tree Accuracy :", round(tree_accuracy, 4))
print("Random Forest Accuracy :", round(forest_accuracy, 4))
if forest_accuracy > tree_accuracy:
    print("\nRandom Forest performed better.")
elif tree_accuracy > forest_accuracy:
    print("\nDecision Tree performed better.")
else:
    print("\nBoth models performed equally.")
```
